```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df=pd.read_csv("Mall_Customers.csv")

print(df.head())
print(df.info())
print(df.isnull().sum())
```

```python
from sklearn.preprocessing import StandardScaler

X=df[['Annual Income (k$)',
      'Spending Score (1-100)']]
scaler=StandardScaler()
X_scaled=scaler.fit_transform(X)
```

```python
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

inertias = []

for k in range(1, 11):
    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )
    model.fit(X_scaled)
    inertias.append(model.inertia_)

plt.figure(figsize=(8, 5))
plt.plot(range(1, 11), inertias, marker='o')
plt.xlabel("Number of Clusters")
plt.ylabel("Inertia")
plt.title("Elbow Method")
plt.xticks(range(1, 11))
plt.grid(True)
plt.show()
```

```python

k = 5

kmeans = KMeans(
    n_clusters=k,
    random_state=42,
    n_init=10
)

kmeans_labels = kmeans.fit_predict(X_scaled)

# Plot customer clusters
plt.figure(figsize=(8, 5))

plt.scatter(
    X_scaled[:, 0],
    X_scaled[:, 1],
    c=kmeans_labels,
    cmap='viridis',
    s=50
)

# Plot cluster centroids
plt.scatter(
    kmeans.cluster_centers_[:, 0],
    kmeans.cluster_centers_[:, 1],
    c='red',
    marker='X',
    s=200,
    label='Centroids'
)

plt.xlabel("Standardized Annual Income")
plt.ylabel("Standardized Spending Score")
plt.title("K-Means Clustering")
plt.legend()
plt.grid(True)
plt.show()
```

```python
from sklearn.cluster import AgglomerativeClustering
import matplotlib.pyplot as plt

# Set the number of clusters
k = 5

# Apply Agglomerative Hierarchical Clustering
hierarchical = AgglomerativeClustering(
    n_clusters=k,
    linkage='ward'
)

hierarchical_labels = hierarchical.fit_predict(X_scaled)

# Visualize the clusters
plt.figure(figsize=(8, 5))

plt.scatter(
    X_scaled[:, 0],
    X_scaled[:, 1],
    c=hierarchical_labels,
    cmap='viridis',
    s=50
)

plt.xlabel("Standardized Annual Income")
plt.ylabel("Standardized Spending Score")
plt.title("Agglomerative Hierarchical Clustering")
plt.grid(True)

plt.show()
```

```python
from scipy.cluster.hierarchy import linkage, dendrogram
import matplotlib.pyplot as plt

# Calculate the linkage matrix
linked = linkage(X_scaled, method='ward')

# Plot the dendrogram
plt.figure(figsize=(12, 5))

dendrogram(linked)

plt.title("Hierarchical Clustering Dendrogram")
plt.xlabel("Customers")
plt.ylabel("Linkage Distance")

plt.tight_layout()
plt.show()
```

```python
from sklearn.metrics import silhouette_score
import numpy as np
import pandas as pd

# 1. Calculate Silhouette Scores
kmeans_sil = silhouette_score(
    X_scaled, kmeans_labels
)

hierarchical_sil = silhouette_score(
    X_scaled, hierarchical_labels
)

# 2. Function to calculate Inertia
def calculate_inertia(X, labels):
    total = 0

    for cluster in np.unique(labels):
        points = X[labels == cluster]
        centroid = points.mean(axis=0)

        total += np.sum(
            (points - centroid) ** 2
        )

    return total

# 3. Calculate Inertia for both algorithms
kmeans_inertia = kmeans.inertia_

hierarchical_inertia = calculate_inertia(
    X_scaled, hierarchical_labels
)

# 4. Create comparison table
results = pd.DataFrame({
    "Algorithm": [
        "K-Means",
        "Agglomerative"
    ],
    "Inertia": [
        kmeans_inertia,
        hierarchical_inertia
    ],
    "Silhouette Score": [
        kmeans_sil,
        hierarchical_sil
    ]
})

# 5. Display results
print("Clustering Evaluation Results:")
print(results.round(4).to_string(index=False))
```
